XRF Cuttings Analysis

X-ray fluorescence (XRF) is becoming a common wellsite cuttings analysis technique (Carr et al, 2014). Portable XRF provides rapid elemental composition measurements of cuttings as they are drilled. These devices measure the fluorescent X-rays emitted by a sample when illuminated by an energetic X-ray source. Elements in the sample will emit X-rays at specific wavelengths, and the spectrum of the emitted X-rays can be measured and used to quantify the amount of the corresponding elements present in the sample. Trends in element concentration can be used to infer sediment depositional environment, sources, and indicate conditions conducive to the preservation of organic material. XRF data can be used for geologic characterization, optimizing well placement, and providing additional guidance for geosteering.

The data for this tutorial consists of XRF measurements of cuttings from the lateral section of an unconventional well. The measurements were made at approximately 10 m intervals. In this case, the raw data consists of 22 measurements for each sample. Each measurement gives the weight percentage of a chemical component. The data is read from a csv file into a dataframe using the pandas library, which provides many convenient data structures and tools for data science.

A dataframe is a 2-dimensional labeled data structure, where each row represents a different sample, and the columns are measurements on the sample. The contents of the dataframe are shown in Table 1.

```python
import os
print("CWD:", os.getcwd())

from pathlib import Path
import pandas as pd

# Tu cwd es TT3/Notebook; parent es TT3/
project_root = Path.cwd().parent  

# Construye la ruta al CSV
csv_path = project_root / 'analysis' / 'data' / 'derived_data' / 'datageo2.csv'

# Carga el DataFrame
geochem_df = pd.read_csv(csv_path)
print(geochem_df.head())
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Extraigo sólo los valores numéricos a escalar
X = geochem_df.iloc[:, 2:]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Devuélvelo a DataFrame si lo necesitas
X_scaled_df = pd.DataFrame(X_scaled,
                           columns=X.columns,
                           index=geochem_df.index)

print(X_scaled_df.head())
```

```python
from sklearn.preprocessing import scale
import numpy as np
import matplotlib.pyplot as plt

# escalado directo en variable 'data'
data = scale( geochem_df.iloc[:, 2:] )

fig = plt.figure(figsize=(10,6))
covar_matrix = np.cov(data, rowvar=False)
eigenvalues = np.linalg.eig(covar_matrix)[0]

plt.plot(eigenvalues, 'o-')
plt.axhline(y=1.0, color='r', linestyle='--')
plt.tick_params(labelsize=18)
plt.xlabel('Eigenvalue Index', fontsize=24)
plt.ylabel('Magnitude', fontsize=24)
plt.show()
```

```python
# 1) Escala tus datos (array numpy)
data = scale(geochem_df.iloc[:, 2:])

# 2) Ajusta el modelo
fa_model = FactorAnalysis(n_components=7)
fa_model.fit(data)

# 3) Extrae los scores (opcional)
factor_data = fa_model.transform(data)
factor_df = pd.DataFrame(factor_data,
                         columns=[f'Comp {i}' for i in range(7)])
factor_df[['Well Name','Depth']] = geochem_df[['Well Name','Depth']]

# 4) Crea el DataFrame de loadings
loading = fa_model.components_  
component_names = geochem_df.columns.values[2:]
# Pon índices claros para cada factor
factor_labels = [f'Factor {i}' for i in range(7)]
loading_df = pd.DataFrame(loading,
                          columns=component_names,
                          index=factor_labels)

# 5) Grafica el segundo factor (índice 1)
factor_id = 1
fig = plt.figure(figsize=(10,4))
plt.ylabel('Factor loadings', fontsize=24)
# <-- aquí reemplazamos .ix[factor_id] por .iloc[factor_id] -->
loading_df.iloc[factor_id].plot(kind='bar')
plt.tick_params(labelsize=20)
plt.title(f'Loadings de {factor_labels[factor_id]}', fontsize=20)
plt.tight_layout()
plt.show()
```

```python

# Extract only the numerical columns and scale them
data = scale(geochem_df.iloc[:, 2:])

# Fit the Factor Analysis model
fa_model = FactorAnalysis(n_components=7, random_state=0)
fa_model.fit(data)

# Get the factor scores
factor_data = fa_model.transform(data)
factor_df = pd.DataFrame(
    factor_data,
    columns=[f'Comp {i}' for i in range(7)]
)
# Add metadata columns
factor_df['Well Name'] = geochem_df['Well Name']
factor_df['Depth']     = geochem_df['Depth']
```

```python
inertias = []
ks = range(1, 30)  # try k = 1 to 29

for k in ks:
    km = KMeans(n_clusters=k, random_state=0).fit(factor_data)
    inertias.append(km.inertia_)

fig = plt.figure(figsize=(10, 5))
plt.plot(ks, inertias, 'o-', linewidth=2)
plt.xlabel('Number of clusters (k)', fontsize=24)
plt.ylabel('Sum of squared errors (inertia)', fontsize=24)
plt.xticks(ks, fontsize=18)
plt.yticks(fontsize=18)
plt.title('Elbow Method for K-Means', fontsize=20)
plt.tight_layout()
plt.show()# fig.savefig('images/Figure_3_k_means.png', dpi=300, bbox_inches='tight')
```

```python

# Each row = a factor, each column = a variable
loading = fa_model.components_
component_names = geochem_df.columns.values[2:]
factor_labels = [f'Factor {i}' for i in range(7)]

loading_df = pd.DataFrame(
    loading,
    index=factor_labels,
    columns=component_names
)

# Example: display loadings for Factor 1
print(loading_df.loc['Factor 1'])
```

```python
# 1. Instantiate and fit the model on your factor scores
kmeans = KMeans(n_clusters=10, random_state=0)
kmeans.fit(factor_data)

# 2. Extract the raw labels (0–9) and, if you prefer, shift to 1–10 for readability
raw_labels = kmeans.labels_
one_based_labels = raw_labels + 1

# 3. Add the cluster assignments back into your DataFrames
#    - factor_df holds the factor scores + metadata
#    - geochem_df holds the original variables + metadata
factor_df = factor_df.assign(Cluster=one_based_labels)
geochem_df = geochem_df.assign(Cluster=raw_labels)

# Now you can inspect how wells/depths fall into each cluster:
print(factor_df.head())
print(geochem_df[['Well Name', 'Depth', 'Cluster']].head())
```

```python
facies_colors = sns.color_palette("Set3", 10)

# dynamically pick all your diatom-cols (everything except meta)
vars_to_plot = [c for c in geochem_df.columns 
                if c not in ['Well Name','Depth','Cluster']]

n = len(vars_to_plot)
ncols = 4
nrows = (n + ncols - 1) // ncols

fig, axes = plt.subplots(nrows, ncols,
                         figsize=(5*ncols, 4*nrows),
                         sharex=True)

for ax, var in zip(axes.flatten(), vars_to_plot):
    # pass hue='Cluster', disable dodge so boxes don’t split,
    # supply your palette, then drop the legend
    sns.boxplot(
        x='Cluster',
        y=var,
        hue='Cluster',
        data=geochem_df,
        palette=facies_colors,
        dodge=False,
        linewidth=0.3,
        ax=ax
    )
    ax.set_xlabel("Cluster Index", fontsize=14)
    ax.set_ylabel(var, fontsize=14)
    ax.tick_params(labelsize=10)
    # remove the automatic legend
    ax.get_legend().remove()

# hide any unused axes
for ax in axes.flatten()[n:]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import colors
from matplotlib.colors import BoundaryNorm
from mpl_toolkits.axes_grid1 import make_axes_locatable
import matplotlib as mpl

def make_facies_log_plot(logs, facies_colors, label):
    # 0) reset estilo matplotlib
    mpl.rcParams.update(mpl.rcParamsDefault)
    
    # 1) ordenar por profundidad y fijar rango manual
    logs = logs.sort_values('Depth')
    ztop, zbot = 38, 266   # <— depth must run from 38 down to 266
    
    # 2) preparar cmap y norm discretos
    clusters = sorted(logs[label].unique())
    cats    = pd.Categorical(logs[label], categories=clusters)
    codes   = cats.codes
    cmap_facies = colors.ListedColormap(facies_colors[:len(clusters)], name='indexed')
    norm        = BoundaryNorm(np.arange(len(clusters)+1)-0.5, len(clusters))
    
    # 3) expandir códigos en matriz para imshow
    cluster = np.repeat(codes[:, None], 100, axis=1)
    
    # 4) crear 13 paneles: 12 curvas + 1 franja
    fig, ax = plt.subplots(1, 13, figsize=(14*1.2, 13), sharey=True)
    ax = np.array(ax).ravel()
    
    # 5) dibujar las 12 curvas geoquímicas
    vars_to_plot = ['TC','TIC','TOC','TN','TS',
                    'd13C','d18O','EM1','EM2','EM3','EM4','EM5']
    for i, var in enumerate(vars_to_plot):
        ax[i].plot(logs[var], logs.Depth, '-', color=f'C{i%10}')
        ax[i].set_xlim(logs[var].min(), logs[var].max())
        ax[i].set_xlabel(var, fontsize=14)
        ax[i].set_ylim(ztop, zbot)
        ax[i].invert_yaxis()
        ax[i].tick_params(labelsize=10)
        if i > 0:
            ax[i].set_yticklabels([])
    
    # 6) dibujar la franja de facies en ax[12], estirada de 38 a 266
    im = ax[12].imshow(
        cluster,
        interpolation='none',
        aspect='auto',
        cmap=cmap_facies,
        norm=norm,
        origin='upper',
        extent=[0, 1, ztop, zbot]   # <— cover full depth range
    )
    divider = make_axes_locatable(ax[12])
    cax     = divider.append_axes("right", size="20%", pad=0.5)
    cbar    = plt.colorbar(im, cax=cax, boundaries=np.arange(len(clusters)+3)-0.1)
    cbar.set_ticks(np.arange(len(clusters)))
    cbar.set_ticklabels([f'Clust {c}' for c in clusters], fontsize=12)
    cbar.set_label('', fontsize=14)
    ax[12].set_xticks([]); ax[12].set_yticklabels([])
    ax[12].set_ylim(ztop, zbot); ax[12].invert_yaxis()
    ax[12].set_xlabel('Facies', fontsize=14)
    
    # 7) poner escala de profundidad de 38 a 266 en izquierda y derecha
    ticks = np.linspace(ztop, zbot, 8)  # e.g. 8 ticks; ajusta si quieres otro número
    tick_labels = [f'{int(t)}' for t in ticks]
    
    # izquierda
    ax[0].set_ylabel('Depth [cm]', fontsize=14)
    ax[0].set_yticks(ticks)
    ax[0].set_yticklabels(tick_labels, fontsize=12)
    
    # derecha junto a la franja
    ax[12].yaxis.set_ticks_position('right')
    ax[12].yaxis.set_label_position('right')
    ax[12].set_yticks(ticks)
    ax[12].set_yticklabels(tick_labels, fontsize=12)
    ax[12].set_ylabel('', fontsize=14)
    
    plt.tight_layout()
    fig.tight_layout(pad=2.0)               # espacio extra alrededor
    fig.subplots_adjust(wspace=0.3,         # separación horizontal
                        left=0.03,          # margen izquierdo
                        right=0.97)         # margen derecho

    return fig
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1) Prepara tu paleta (tantos colores como clusters tengas)
facies_colors = sns.color_palette("Set3", 10)

# 2) Llama a la función; te devuelve el objeto Figure
fig = make_facies_log_plot(
    logs=geochem_df,       # tu DataFrame con Depth, Cluster y proxies
    facies_colors=facies_colors,
    label='Cluster'         # el nombre de la columna que indica facies/clusters
)

# 3) Muestras la ventana con todos los ejes
plt.show()

# (Opcional) Guardar en disco
# fig.savefig('facies_log_plot.png', dpi=300, bbox_inches='tight')
```
