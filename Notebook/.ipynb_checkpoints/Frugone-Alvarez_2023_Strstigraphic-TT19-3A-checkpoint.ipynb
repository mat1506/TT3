XRF Cuttings Analysis

X-ray fluorescence (XRF) is becoming a common wellsite cuttings analysis technique (Carr et al, 2014). Portable XRF provides rapid elemental composition measurements of cuttings as they are drilled. These devices measure the fluorescent X-rays emitted by a sample when illuminated by an energetic X-ray source. Elements in the sample will emit X-rays at specific wavelengths, and the spectrum of the emitted X-rays can be measured and used to quantify the amount of the corresponding elements present in the sample. Trends in element concentration can be used to infer sediment depositional environment, sources, and indicate conditions conducive to the preservation of organic material. XRF data can be used for geologic characterization, optimizing well placement, and providing additional guidance for geosteering.

The data for this tutorial consists of XRF measurements of cuttings from the lateral section of an unconventional well. The measurements were made at approximately 10 m intervals. In this case, the raw data consists of 22 measurements for each sample. Each measurement gives the weight percentage of a chemical component. The data is read from a csv file into a dataframe using the pandas library, which provides many convenient data structures and tools for data science.

A dataframe is a 2-dimensional labeled data structure, where each row represents a different sample, and the columns are measurements on the sample. The contents of the dataframe are shown in Table 1.

```python
import pandas as pd
geochem_df = pd.read_csv('/home/ma1506/Dropbox/Villascis_et_al_2019/DATOS/DIATOMEAS/DIATOM_dataset2.csv')
geochem_df
```

```python
geochem_df['Si/Ti'] = geochem_df['Si'] / geochem_df['Ti']
geochem_df['Rb/Sr'] = geochem_df['Rb'] / geochem_df['Sr']
geochem_df['S/Ti'] = geochem_df['S'] / geochem_df['Ti']
geochem_df['Br/Ti'] = geochem_df['Br'] / geochem_df['Ti']
geochem_df['Fe/Mn'] = geochem_df['Fe'] / geochem_df['Mn']
```

```python
from sklearn.preprocessing import scale
data = geochem_df.ix[:, 2:]
data = scale(data)
```

```python
fig = plt.figure(figsize=(10,6))
covar_matrix = np.cov(data, rowvar=False)
eigenvalues = np.linalg.eig(covar_matrix)[0]
plt.plot(eigenvalues, 'o-')
plt.axhline(y=1.0, color='r', linestyle='--')
plt.tick_params(labelsize=18)
plt.xlabel('Eigenvalue Index', fontsize=24)
plt.ylabel('Magnitude', fontsize=24)
# fig.savefig('images/Figure_1_eigs.tif', dpi=300, bbox_inches='tight')
# fig.savefig('images/Figure_1_eigs.png', dpi=300, bbox_inches='tight')
```

```python
fa_model = FactorAnalysis(n_components = 7)
fa_model.fit(data)
factor_data = fa_model.transform(data)
```

```python
# create a dataframe consisting of the factor values for each data point
factor_df = pd.DataFrame(factor_data, columns=['Comp {}'.format(i) for i in range(0,7)])
factor_df['Well Name'] = geochem_df['Well Name']
factor_df['Depth'] = geochem_df['Depth']

# create a dataframe with the factor loadings associated with each observation
loading = fa_model.components_
component_names = geochem_df.columns.values[2:]
loading_df =  pd.DataFrame(loading,
                           columns=component_names)
```

```python
for factor_id in range(0, loading_df.shape[0]):

    # just plot the fourth factor, remove the loop to plot all
    if factor_id == 1:
        fig = plt.figure(figsize=(10,4))
        #plt.title('Factor {}'.format(factor_id))
        plt.ylabel('Factor loadings', fontsize=24)
        loading_df.ix[factor_id].plot(kind='bar')
        plt.tick_params(labelsize=20)
        # fig.savefig('images/Figure_2_Factor_interp.png', dpi=300, bbox_inches='tight')
        # fig.savefig('images/Figure_2_Factor_interp.tif', dpi=300, bbox_inches='tight')
        break
    else:
        continue
```

```python
inertias = []
means = []
maxK = 30
for k in range(1, maxK):
    kmeans = KMeans(n_clusters=k, random_state=0).fit(factor_data)
    means.append(k)
    inertias.append(kmeans.inertia_)

fig = plt.figure(figsize=(10,5))
plt.plot(means, inertias, 'o-')
plt.xlabel('# of clusters', fontsize=24)
plt.ylabel('Sum of squared errors', fontsize=24)
plt.tick_params(labelsize=18)
# fig.savefig('images/Figure_3_k_means.tif', dpi=300, bbox_inches='tight')
# fig.savefig('images/Figure_3_k_means.png', dpi=300, bbox_inches='tight')
```

```python
kmeans = KMeans(n_clusters=10, random_state=0)
kmeans.fit(factor_data)

# add the cluster ids to the dataset
factor_df['Cluster'] = kmeans.labels_ + 1
geochem_df['Cluster'] = kmeans.labels_
```

```python
def make_facies_log_plot(logs, facies_colors, label):
    # reset to original matplotlib style
    # seaborn doesn't look as good for this
    import matplotlib as mpl
    mpl.rcParams.update(mpl.rcParamsDefault)
    
    #make sure logs are sorted by depth 
    logs = logs.sort_values(by='Depth')
    cmap_facies = colors.ListedColormap(
            facies_colors[0:len(facies_colors)], 'indexed')
    
    ztop=logs.Depth.min(); zbot=logs.Depth.max()
    
    cluster=np.repeat(np.expand_dims(logs[label].values,1), 100, 1)
    
    f, ax = plt.subplots(nrows=1, ncols=11, figsize=(10, 12))
    ax[0].plot(logs['Cocconeis placentula'], logs.Depth, '-g')
    ax[1].plot(logs['Tabularia fasciculata'], logs.Depth, '-')
    ax[2].plot(logs['Melosira varians'], logs.Depth, '-', color='0.5')
    ax[3].plot(logs['Diploneis didyma'], logs.Depth, '-', color='r')
    ax[4].plot(logs['P/B'], logs.Depth, '-', color='r')
    ax[5].plot(logs['M/F'], logs.Depth, '-', color='r')
    ax[6].plot(logs['Cyclotella meneghiniana'], logs.Depth, '-', color='r')
    ax[7].plot(logs['Hyalodiscus spp'], logs.Depth, '-', color='r')
    ax[8].plot(logs['Cocconeis placentula'], logs.Depth, '-', color='r')
    ax[9].plot(logs['Aulacoseira granulata'], logs.Depth, '-', color='r')
    im=ax[10].imshow(cluster, interpolation='none', aspect='auto',cmap=cmap_facies,vmin=0,vmax=10)
    
    divider = make_axes_locatable(ax[10])
    cax = divider.append_axes("right", size="20%", pad=0.05)
    cbar=plt.colorbar(im, cax=cax)
    cbar.set_label((10*' ').join(['Clust 1', 'Clust 2', 'Clust 3', 
                                'Clust 4', 'Clust 5', 'Clust 6','Clust 7','Clust 8','Clust 9','Clust 10']), fontsize=14)
    cbar.set_ticks(range(0,1)); cbar.set_ticklabels('')
    
    
    for i in range(len(ax)-1):
        ax[i].set_ylim(ztop,zbot)
        ax[i].invert_yaxis()
        ax[i].grid()
        ax[i].locator_params(axis='x', nbins=3)
    
    ax[0].set_xlabel('Cocconeis placentula', fontsize=16)
    ax[0].set_xlim(logs['Cocconeis placentula'].min(),logs['Cocconeis placentula'].max())
    ax[0].set_ylabel('Depth [mm]', fontsize=24)
    ax[0].tick_params(labelsize=12)
    ax[0].grid(b=False)
    ax[1].set_xlabel("Tabularia fasciculata", fontsize=16)
    ax[1].set_xlim(logs['Tabularia fasciculata'].min(),logs['Tabularia fasciculata'].max())
    ax[1].grid(b=False)
    ax[1].tick_params(labelsize=12)
    ax[2].set_xlabel("Melosira varians", fontsize=16)
    ax[2].set_xlim(logs['Melosira varians'].min(),logs['Melosira varians'].max())
    ax[2].grid(b=False)
    ax[2].tick_params(labelsize=12)
    ax[3].set_xlabel("Diploneis didyma", fontsize=16)
    ax[3].set_xlim(logs['Diploneis didyma'].min(),logs['Diploneis didyma'].max())
    ax[3].grid(b=False)
    ax[3].tick_params(labelsize=12)
    ax[4].set_xlabel("P/B", fontsize=16)
    ax[4].set_xlim(logs['P/B'].min(),logs['P/B'].max())
    ax[4].grid(b=False)
    ax[4].tick_params(labelsize=12)
    ax[5].set_xlabel("M/F", fontsize=16)
    ax[5].set_xlim(logs['M/F'].min(),logs['M/F'].max())
    ax[5].grid(b=False)
    ax[5].tick_params(labelsize=12)
    ax[6].set_xlabel("Cyclotella meneghiniana", fontsize=16)
    ax[6].set_xlim(logs['Cyclotella meneghiniana'].min(),logs['Cyclotella meneghiniana'].max())
    ax[6].grid(b=False)
    ax[6].tick_params(labelsize=12)
    ax[7].set_xlabel("Hyalodiscus spp", fontsize=16)
    ax[7].set_xlim(logs['Hyalodiscus spp'].min(),logs['Hyalodiscus spp'].max())
    ax[7].grid(b=False)
    ax[7].tick_params(labelsize=12)
    ax[8].set_xlabel("Cocconeis placentula", fontsize=16)
    ax[8].set_xlim(logs['Cocconeis placentula'].min(),logs['Cocconeis placentula'].max())
    ax[8].grid(b=False)
    ax[8].tick_params(labelsize=12)
    ax[9].set_xlabel("Aulacoseira granulata", fontsize=16)
    ax[9].set_xlim(logs['Aulacoseira granulata'].min(),logs['Aulacoseira granulata'].max())
    ax[9].grid(b=False)
    ax[9].tick_params(labelsize=12)
    ax[10].set_xlabel('Facies', fontsize=16)
    
    ax[1].set_yticklabels([]); ax[2].set_yticklabels([]); ax[3].set_yticklabels([]); ax[4].set_yticklabels([]); ax[5].set_yticklabels([]); ax[6].set_yticklabels([]); ax[7].set_yticklabels([]); ax[8].set_yticklabels([]); ax[9].set_yticklabels([])
    ax[10].set_yticklabels([]);
    ax[10].set_xticklabels([])
    
    #f.suptitle(logs.iloc[0]['Well Name'], fontsize=14,y=0.9)
    return f
```
